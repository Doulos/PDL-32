### Dataset Generation

```python
%matplotlib inline
import numpy as np
import matplotlib.pyplot as plt

# Generate dataset

m = 200
train_slope  = 0.1
train_offset = -1.0

train_x = np.linspace(-10, 10, m).astype(np.float32)
rng = np.random.RandomState(seed=42)
train_y = (train_slope * train_x + np.sin(train_x / 1.5) + np.cos(train_x / 0.7) + train_offset +
           rng.normal(0.0, 0.2, size=len(train_x))).astype(np.float32)

plt.figure(1, figsize=(12, 8))
plt.plot(train_x, train_y, 'o');
plt.show()
```

### Data Preparation

Reshape the data into 2-D arrays as before, then normalize the data to bring it into the range [-1.0,+1.0]

```python
train_x = train_x.reshape(-1, 1)
train_y = train_y.reshape(-1, 1)
train_x = train_x - np.mean(train_x, axis=0)
train_x = train_x / np.max(train_x, axis=0)
```

### Neural Network Model
Implement a neural network and gradient descent algorithm for non-linear regression in TensorFlow.  As a starting point, the cell below implement the model we used on Lab4 for the linear regression. As you will see, it is quite easy to modify the model to deal with non-linear regression. 

Compare the performance of networks with one and with two hidden layers and with various numbers of hidden units - perhaps two layers of 16 or 32 hidden units, each with ReLU activation, plus the output layer as shown on the slides. As usual, you will need to adjust some model hyper-parameters: number of training steps (`n_steps`) and learning rate (`learning_rate`). You my also try using `'adam'` as optimizer, instead of SGD.

Feel free to experiments!

```python
import tensorflow
from tensorflow.keras.models import Sequential
from tensorflow.keras.layers import Dense,Input
from tensorflow.keras.optimizers import SGD

tensorflow.keras.backend.clear_session()

#----- this is our linear regression model from Ex04. You need to adjust it!

model = Sequential()
model.add(Input(shape=(1,)))
model.add(Dense(units=1))
model.summary()

model.compile(loss='mean_squared_error', optimizer=SGD(learning_rate=0.1))

n_steps = 300
n_iter  = 10

#-----

# Split the training in n_iters parts
n_steps_per_iter = n_steps//n_iter
for step in range(n_steps_per_iter,n_steps+1,n_steps_per_iter):
    model.fit(train_x, train_y, epochs=n_steps_per_iter, batch_size=m, verbose=0)

    # checkpoint: show current model performance
    loss = model.evaluate(train_x, train_y, batch_size=m, verbose=0)
    print(f'Loss at step {step:5} = {loss:5.3f}')

prediction = model.predict(train_x, batch_size=m)

plt.figure(1, figsize=(12, 8))
plt.plot(train_x, prediction)
plt.plot(train_x, train_y, 'o');
plt.show()
```
