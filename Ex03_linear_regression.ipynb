### Stochastic Gradient Descent

Now copy and modify your code from above to implement a stochastic gradient descent algorithm.

Once again print out the cost at each step and show the scatter plot of slope and offset during stochastic gradient descent. You may need to experiment with the initial values of the slope and offset variables to get a good plot.

```python
def hypothesis(x, slope, offset):
    return slope * x + offset

def cost(x, slope, offset, y):
    return np.mean(np.square(hypothesis(x, slope, offset) - y)) / 2.0

def derivative_wrt_slope(x, slope, offset, y):
    return np.mean((hypothesis(x, slope, offset) - y) * x)

def derivative_wrt_offset(x, slope, offset, y):
    return np.mean((hypothesis(x, slope, offset) - y))


# num_steps and learning_rate
num_steps = 50
learning_rate = 0.1

# initial slope/offset
slope = +4.0
offset = -5.0

# descent gradient algorithm
slope_at_step  = np.empty((num_steps))
offset_at_step = np.empty((num_steps))
plt.figure(1, figsize=(12, 8))

for step in range(num_steps):
    # Update learned parameter values
    new_slope  = slope  - learning_rate * derivative_wrt_slope(train_x, slope, offset, train_y)
    new_offset = offset - learning_rate * derivative_wrt_offset(train_x, slope, offset, train_y)
    slope  = new_slope
    offset = new_offset

    slope_at_step[step] = slope
    offset_at_step[step] = offset

    if (step%20==0):
        plt.plot(train_x, slope * train_x + offset)
        print(f'Cost at step {step} = {cost(train_x, slope, offset, train_y):5.3f}')

plt.plot(train_x, train_y, 'o');
plt.plot(train_x, train_y, 'o');
plt.show()

plt.figure(1, figsize=(12, 8))
plt.scatter(offset_at_step, slope_at_step)
plt.show()
print(f"Final {slope=} (ideal: {train_slope}),  {offset=} (ideal: {train_offset})")
```
