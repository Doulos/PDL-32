```python
train_x = train_x.reshape(m,1)
train_y = train_y.reshape(m,1)

import tensorflow
from tensorflow.keras.models import Sequential
from tensorflow.keras.layers import Dense,Input
from tensorflow.keras.optimizers import SGD

tensorflow.keras.backend.clear_session()

model = Sequential()
model.add(Input(shape=(1,)))
model.add(Dense(units=1))
model.summary()

model.compile(loss='mean_squared_error', optimizer=SGD(learning_rate=0.05))

model.fit(train_x, train_y, epochs=300, batch_size=m, verbose=0)

weights = model.get_weights()

print ('The weights of the model are', weights)
```

#### Using KERAS_BACKEND

When you set the **KERAS_BACKEND** parameter to ***torch*** Keras creates a model that uses **torch.nn.Module**.  This is verified by checking that the generated Keras module is an instance of torch.nn.Module.

```python
import os
import keras
from keras import layers
from keras.models import Sequential
from keras.layers import Dense, Input
import torch

# Define a simple sequential model for linear regression
def create_model():
  model = Sequential()
  model.add(Input(shape=(1,)))
  model.add(Dense(units=1))
  return model

keras_model = create_model()

# Compile the model
keras_model.compile(optimizer='adam',
              loss='mean_squared_error') # Using mean squared error for regression

# Show that the underlying module is a PyTorch module
print('The generated Keras module is an instance of torch.nn.Module:', isinstance(keras_model, torch.nn.Module))
keras_model.summary()
```
