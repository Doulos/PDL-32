Download a single image that can be used as a simple check that inference is working

```python
import numpy as np
import pylab
import PIL
from IPython import display
!/usr/bin/wget https://upload.wikimedia.org/wikipedia/commons/f/fe/Giant_Panda_in_Beijing_Zoo_1.JPG -O panda.jpg
img = np.array(PIL.Image.open('panda.jpg').resize((224, 224))).astype(np.float32) / 128 - 1
display.display(display.Image('panda.jpg'))
image = img.reshape(1,224,224,3)
```

Run inference on a single image by calling the eval() method of the model output, passing the image as an argument.

```python
from tensorflow.keras.applications import MobileNetV2

model = MobileNetV2(weights='imagenet', include_top=True)

x = model.predict(image)
print(x.shape)
```

#### What does the shape  of the output tensor tell you about the number of classes the model is working with?

Clone the TensorFlow models GitHub repository just to get human-readable labels for the images in the ImageNet dataset

Print the label belonging to the top prediction from the model

```python
import tensorflow as tf
labels_path = tf.keras.utils.get_file('ImageNetLabels.txt','https://storage.googleapis.com/download.tensorflow.org/data/ImageNetLabels.txt')
label_map = np.array(open(labels_path).read().splitlines())

print("Top 1 Prediction: ", x.argmax(),label_map[x.argmax()], x.max())
```

    import tensorflow
    from tensorflow.keras.applications.mobilenet_v2 import MobileNetV2

    # Default arguments
    model = MobileNetV2()
    x = model.predict(image)

    labels_path =     tf.keras.utils.get_file('ImageNetLabels.txt','https://storage.googleapis.com/download.tensorflow.org/data/ImageNetLabels.txt')
    label_map = np.array(open(labels_path).read().splitlines())

    prediction = x.argmax() + 1
    print("Top 1 Prediction: ", prediction,label_map[prediction], x.max())

    # Explicit arguments
    model = MobileNetV2(input_shape=(128,128,3), alpha=0.5, weights='imagenet', include_top=True)

    img = np.array(PIL.Image.open('panda.jpg').resize((128, 128))).astype(np.float32) / 128 - 1
    image = img.reshape(1,128,128,3)

    x = model.predict(image)

    prediction = x.argmax() + 1
    print("Top 1 Prediction: ", prediction,label_map[prediction], x.max())
