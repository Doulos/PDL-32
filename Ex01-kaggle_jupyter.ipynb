### **Step 3.** Use of Cell Magics Commands
Colab cells can execute bash shell commands using [Cell Magics](https://ipython.readthedocs.io/en/stable/interactive/magics.html) commands. The first command **%%bash** interprets subsequent commands as shell commands

```python
%%bash

ls -l
pwd
```

- Code written as a Python script can be written and saved. Python code is first edited and saved to a file using ***%%writefile*** cell magics command. Subsequently the saved (fibonacci.py) file is  executed the command line Python interpreter.
- The saved Python script file can be viewed using the Folder icon on the left hand panel or using the Terminal.

```python
%%writefile fibonacci.py

def fibonacci_loop(num_terms):
    a, b = 0, 1
    print("Fibonacci sequence:")
    for _ in range(num_terms):
        a, b = b, a + b
        print(a, end=" ")

# Calling the fibonacci function
fibonacci_loop(10)

```

- Colab is a Linux based machine.  You can now execute the saved Python script like you would on a Linux command line.
- A Bash command can also be executed by appending ***!*** symbol.

```python
!python fibonacci.py
```
