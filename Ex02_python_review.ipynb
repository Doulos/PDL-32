### Generators

A Python generator is a convenient way to create a Python iterator. An iterator is an object on which you can call the `next()` function to return the next value in a series, which could be either the next value from an existing collection of values or a new value generated on-the-fly.

Syntactically, a Python generator is a function that calls the `yield` statement instead of the `return` statement.

```python
def gen():
    count = 0
    while (True):
        count += 1
        yield count
```

When the generator function is called, it returns an instance object that can be used as an iterator

```python
iterator = gen()
```

Each call to `next` causes the generator function to wake up from the point where it was paused (immediately after `yield`) and execute through to the next `yield` statement. The value returned by `next` is the value returned by `yield`.

```python
next(iterator)
```

```python
next(iterator)
```

### f-strings

F-strings can be used to embed the values of variables or expressions inside a string. Each expression is written with a placeholder { }. Within a placeholder, {:N} indicates a minimum field width.

```python
for step in range(4):
    loss = 1.0 / (step + 1)
    print(f'Loss at step {(step+1)*300:4} = {loss:5.3f}')
```

Within a placeholder such as `{:N}`, the characters `<`, `^`, and `>` can be used to left, center and right-justify a value within a field of N characters, respectively.

```python
for i in [10000, 1000, 100, 10, 1]:
    print(f'{i:<7} {i:^7} {i:>7}')
```

The Numpy package provides many ways of manipulating Numpy arrays. For example, the function np.random.shuffle() shuffles the elements of an array.

```python
y = np.arange(4)
np.random.shuffle(y)
y
```

It is possible to add an extra dimension to a Numpy array by adding an extra index with the special value `None`. In this case, by adding an extra dimension, we are in effect turning a row vector into a  column vector.

```python
y[:,None]
```

It is possible to combine Numpy arrays of different shapes in a single operation known as a broadcast. Here we are combining a row vector of 4 items with a column vector of 4 items using the == operator. The row vector is replicated vertically and the column vector is replicated horizonally across the entire 4x4 matrix, and the two 4x4 matrices are then compared elementwise to produce a new 4x4 matrix contain a boolean at each position.

```python
np.arange(4) == y[:,None]
```

We can create a one-hot representation of the original shuffled row vector y by converting the elements of this matrix from boolean to floating-point.

```python
y_1_hot = (np.arange(4) == y[:,None]).astype(np.float32)
y_1_hot
```

### List Operations

The Python language construct shown below is known as a list comprehension. It provides a means of creating a list by, in effect, embedding a `for` loop within a list. The function np.random.random() returns a random floating-point value in the range 0.0 to 1.0.

```python
import numpy as np
m = 10
data = [np.random.random() for i in range(m)]
data
```

A list comprehension can contain several nested `for` and `if` statements

```python
triplets = [(i,j,k) for i in range(3) for j in range(3) for k in range(3) if not (i == j and j == k)]
triplets
```

The Python language construct `[]` is a way of writing an empty list. Python provides many methods for manipulating lists, such as the `append` method, which appends one item to a list.

```python
c = []    # Empty list

for i in range(m):
    if data[i] < 0.333333:
        c.append('r')
    elif data[i] < 0.666667:
        c.append('g')
    else:
        c.append('b')

c
```

### Classes

A Python class usually contains functions and variables. The function named `__init__` is the constructor, a function that is called to initialize any instance objects of this class as they are created.

```python
class C:
    def __init__(self):
        self.name = 'default'

    def set_name(self, name):
        self.name = name

    def get_name(self):
        return self.name
```

Calling the class name as if it were a function will return an instance object of that class.

```python
obj1 = C()
obj2 = C()
obj3 = C()
```

Functions that are defined within the class can be called as methods of the instance object. Any functions that are to be called in this way (using the syntax `object.method()`) require an extra argument, which is always the first argument in the function definition, and is conventionally named `self`. When a method is called (`set_name` or `get_name` in this example), the `self` argument is bound to the instance object, which makes any variables associated with the instance object visible inside the functions within the class.

```python
obj1.set_name('Monty')
obj2.set_name('Python')
```

```python
obj1.get_name()
```

```python
obj2.get_name()
```

```python
obj3.get_name()
```

A class can be derived from an existing class, in which case the new class inherits all the attributes of the existing class.

```python
class D(C):
    def __init__(self):
        C.__init__(self)
        self.value = 0

    def set_value(self, value):
        self.value = value

    def get_value(self):
        return self.value
```

```python
obj = D()
```

```python
obj.get_name()  # get_name is inherited from class C
```

```python
obj.get_value() # get_value is defined explicitly in class D
```
